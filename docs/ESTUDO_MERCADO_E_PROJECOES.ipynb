# Synapsis Clínico • Modelo Financeiro e Dimensionamento de Mercado
## Simulação de Faturamento Recorrente (MRR e ARR) para 2% a 5% do Mercado Clínico Brasileiro

Este notebook executa o modelo econométrico de receita recorrente para o **Synapsis Clínico** considerando:
1. Base de **470.000 profissionais clínicos ativos** no Brasil (CFP, CFFa, COFFITO e ABPp);
2. Proporção de mercado entre autônomos Solo (60%), Parcerias (20%) e Clínicas Multidisciplinares (20%);
3. Nova grade oficial de preços do ecossistema.

```python
import pandas as pd

# Parâmetros do Mercado Clínico Brasileiro (SAM)
TOTAL_CLINICAL_MARKET = 470_000

# Proporção da Estrutura Operacional
PCT_SOLO = 0.60
PCT_PARCERIA = 0.20
PCT_CLINICA = 0.20

# Preços Oficiais da Assinatura Anual (R$/mês)
PRICE_SOLO = 49.00
PRICE_PARCERIA = 89.00
PRICE_CLINICA_BASE = 179.00
AVG_CLINICA_TICKET = 195.00  # Ponderando clínicas que contratam Packs adicionais de +5 terapeutas

# Média de Profissionais por Conta
PROF_PER_PARCERIA = 2.3
PROF_PER_CLINICA = 6.0

def simular_cenario(market_share_pct):
    total_prof = int(TOTAL_CLINICAL_MARKET * (market_share_pct / 100.0))
    
    solo_prof = int(total_prof * PCT_SOLO)
    parceria_prof = int(total_prof * PCT_PARCERIA)
    clinica_prof = int(total_prof * PCT_CLINICA)
    
    contas_solo = solo_prof
    contas_parceria = int(parceria_prof / PROF_PER_PARCERIA)
    contas_clinica = int(clinica_prof / PROF_PER_CLINICA)
    
    mrr_solo = contas_solo * PRICE_SOLO
    mrr_parceria = contas_parceria * PRICE_PARCERIA
    mrr_clinica = contas_clinica * AVG_CLINICA_TICKET
    
    mrr_total = mrr_solo + mrr_parceria + mrr_clinica
    arr_total = mrr_total * 12
    
    return {
        'Market Share (%)': f"{market_share_pct:.1f}%".replace('.', ','),
        'Profissionais': f"{total_prof:,}".replace(',', '.'),
        'Contas Solo': f"{contas_solo:,}".replace(',', '.'),
        'Contas Parceria': f"{contas_parceria:,}".replace(',', '.'),
        'Clínicas Multi': f"{contas_clinica:,}".replace(',', '.'),
        'MRR Mensal (R$)': f"R$ {mrr_total:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.'),
        'ARR Anual (R$)': f"R$ {arr_total:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')
    }

cenarios = [1.0, 2.0, 3.0, 3.5, 4.0, 5.0, 7.5, 10.0]
df_simulacao = pd.DataFrame([simular_cenario(c) for c in cenarios])
df_simulacao
```
